**Pruning paths** — accuracy vs sparsity per method. LRP beats random at every sparsity; magnitude stays ahead on clean MNIST (an honest nuance).

```
from IPython.display import Image, display
display(Image("artifacts/pruning_curves.png", width="65%"))
```

```
import json, numpy as np, matplotlib.pyplot as plt
rows = json.load(open(f"{ARTIFACTS}/phase5_pruning.json"))
for m in ("lrp", "magnitude", "random"):
    xs = [r["sparsity"] for r in rows if r["method"]==m]
    ys = [r["acc"] for r in rows if r["method"]==m]
    plt.plot(xs, ys, "o-", label=m)
plt.xlabel("sparsity"); plt.ylabel("test accuracy")
plt.legend(); plt.title("Attribution-guided pruning"); plt.show()
```
